# **Data Quality Assessment**

```python
missing_values = df.isnull().sum()

missing_values
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 6))

missing_percentage.sort_values(ascending=False).plot(kind="bar")

plt.title("Percentage of Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Percentage (%)")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The missing-value analysis identifies columns containing incomplete information.

Columns such as `director`, `cast`, and `country` contain a considerable number of missing values.

The `description` column is generally more complete.

Missing values must be handled carefully because removing too many rows could result in significant data loss.

# **Duplicate Records**

```python
df.duplicated().sum()
```

### Interpretation

The duplicate check determines whether the dataset contains repeated records.

If duplicate records are found, they should be removed before performing further analysis.

```python
df = df.drop_duplicates()

df.duplicated().sum()
```

## 🧹 Data Cleaning

The following cleaning steps will be performed:

1. Remove duplicate records.
2. Handle missing values.
3. Convert date columns into appropriate datetime format.
4. Standardize column values where necessary.
5. Prepare the dataset for further analysis.

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
df["date_added"].dtype
```

### Interpretation

The `date_added` column has been converted into datetime format, making it easier to perform time-based analysis.

## Feature Engineering

New features will be created from existing columns to make the analysis easier.

The following features will be created:

- `added_year` — year the content was added to Netflix
- `added_month` — month the content was added

```python
df["added_year"] = df["date_added"].dt.year
df["added_month"] = df["date_added"].dt.month
```

```python
df[["date_added", "added_year", "added_month"]].head(20)
```

### Interpretation

The new features allow us to analyze when Netflix added content to its platform.

This makes year-based and month-based content trend analysis easier.

# **Content Type Analysis**

```python
type_counts = df["type"].value_counts()

type_counts
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="type")

plt.title("Distribution of Netflix Content Types")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

### Interpretation

The analysis compares the number of Movies and TV Shows available in the dataset.

The visualization makes it easy to identify which content type is more common in the Netflix dataset.

# **Movies vs TV Shows Percentage**

```python
type_percentage = df["type"].value_counts(normalize=True) * 100

type_percentage
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Movies vs TV Shows on Netflix")

plt.show()
```

### Interpretation

The percentage distribution provides a clearer comparison between Movies and TV Shows.

This helps us understand the overall composition of Netflix's content library.

# **Content Added Over the Years**

```python
yearly_content = df["added_year"].value_counts().sort_index()

yearly_content
```

```python
plt.figure(figsize=(12, 6))

plt.plot(yearly_content.index, yearly_content.values, marker="o")

plt.title("Netflix Content Added Over the Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.grid(True)
plt.show()
```

### Interpretation

The time-series analysis shows how the number of titles added to Netflix changed over the years.

This helps identify periods when Netflix expanded its content library more rapidly.

# **Release Year Analysis**

```python
release_year_counts = df["release_year"].value_counts().sort_index()

release_year_counts.tail(20)
```

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    data=df,
    x="release_year",
    bins=30
)

plt.title("Distribution of Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

### Interpretation

The release-year distribution shows how old or recent the content in the dataset is.

A large concentration of titles in recent years indicates that the dataset contains a significant amount of relatively modern content.

# **Most Common Ratings**

```python
rating_counts = df["rating"].value_counts()

rating_counts
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="rating",
    order=df["rating"].value_counts().index
)

plt.title("Netflix Content by Rating")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.show()
```

Top Countries

The country column can contain multiple countries in one cell, so we should split it before analyzing individual countries.

```python
country_df = df.assign(
    country=df["country"].str.split(", ")
).explode("country")
```

```python
top_countries = country_df["country"].value_counts().head(10)

top_countries
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

### Interpretation

The country analysis identifies the countries most frequently associated with Netflix content.

Because a title can be associated with multiple countries, the dataset was split and exploded before counting countries individually.

# **Most Common Genres**

```python
genre_df = df.assign(
    genre=df["listed_in"].str.split(", ")
).explode("genre")
```

```python
top_genres = genre_df["genre"].value_counts().head(10)

top_genres
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

### Interpretation

The genre analysis identifies the most frequently occurring categories in the Netflix dataset.

This provides insight into the types of content that are most strongly represented on the platform.

# **Movie Duration Analysis**

```python
movies = df[df["type"] == "Movie"].copy()

movies["duration_minutes"] = (
    movies["duration"]
    .str.extract("(\d+)")
    .astype(float)
)
```

```python
movies["duration_minutes"].describe()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=movies,
    x="duration_minutes",
    bins=30
)

plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

### Interpretation

The movie-duration analysis shows how long Netflix movies typically are.

The distribution can help identify the most common movie-length range and unusually short or long movies.

# **TV Show Seasons Analysis**

```python
tv_shows = df[df["type"] == "TV Show"].copy()

tv_shows["seasons"] = (
    tv_shows["duration"]
    .str.extract("(\d+)")
    .astype(float)
)
```

```python
tv_shows["seasons"].describe()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=tv_shows,
    x="seasons"
)

plt.title("Number of Seasons in Netflix TV Shows")
plt.xlabel("Number of Seasons")
plt.ylabel("Number of TV Shows")

plt.show()
```

### Interpretation

The analysis shows how many seasons Netflix TV shows typically contain.

This allows us to compare the structure of TV content and identify whether short or long-running shows are more common.

# **Movies vs TV Shows by Release Year**

```python
content_by_year = (
    df.groupby(["release_year", "type"])
    .size()
    .reset_index(name="count")
)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=content_by_year,
    x="release_year",
    y="count",
    hue="type"
)

plt.title("Movies vs TV Shows by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

### Interpretation

This comparison shows how Movies and TV Shows are distributed across release years.

It allows us to observe changes in the representation of different content types over time.

# **Top Directors**

```python
top_directors = df["director"].value_counts().head(10)

top_directors
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_directors.values,
    y=top_directors.index
)

plt.title("Top 10 Directors by Number of Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Director")

plt.show()
```

### Interpretation

The analysis identifies directors associated with the largest number of titles in the dataset.

Missing director information means this analysis only considers records where a director is available.
